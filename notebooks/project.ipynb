```python
# import libraries
import pandas as pd

# load csvs as dfs
df_bcpi = pd.read_csv("/Users/abijeetdhillon/MDSA/DATA604/data604_project/data/raw/BCPI_MONTHLY-sd-1972-01-01.csv")
df_ceer = pd.read_csv("/Users/abijeetdhillon/MDSA/DATA604/data604_project/data/raw/CEER_BROADM,CEER_MCM,CEER_OITPM,CEER_BROAD_XUSM-sd-1990-01-01.csv")
```

```python
# check if duplicates
duplicates_bcpi = df_bcpi[df_bcpi['date'].duplicated()]
duplicated_ceer = df_ceer[df_ceer['date'].duplicated()]
print(len(duplicates_bcpi))
print(len(duplicated_ceer))

# convert date string column to date datetime column type
df_bcpi['date'] = pd.to_datetime(df_bcpi['date'])
df_ceer['date'] = pd.to_datetime(df_ceer['date'])

# change datetime format to yyyy-mm
df_bcpi['date'] = pd.to_datetime(df_bcpi['date']).dt.strftime("%Y-%m")
df_ceer['date'] = pd.to_datetime(df_ceer['date']).dt.strftime("%Y-%m")

# make both dataframes only have data from jan 2015 to dec 2025
df_bcpi = df_bcpi.loc[(df_bcpi['date'] >= "2015-01") & (df_bcpi['date'] <= '2025-12')]
df_ceer = df_ceer.loc[(df_ceer['date'] >= "2015-01") & (df_ceer['date'] <= '2025-12')]
display(df_bcpi)
display(df_ceer)
```

```python
# create combined dataset
df_combined = df_bcpi.merge(df_ceer, on = "date", how = "outer")
df_combined['geography'] = "Canada"
display(df_combined)
```

```python
# save combined dataset to /data/processed
from pathlib import Path
import os

combined_dir_path = Path("../data/processed/")
combined_dir_path.mkdir(parents=True, exist_ok=True)

combined_file_path = combined_dir_path / "df_combined.csv"
df_combined.to_csv(combined_file_path)
```
